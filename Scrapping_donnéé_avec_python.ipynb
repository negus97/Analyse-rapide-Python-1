```python
import pandas as pd
import requests
from bs4 import BeautifulSoup

# URL de la page spécifique (ex: PIB de la Côte d'Ivoire)
url = "https://fr.countryeconomy.com/pays/cote-d-ivoire"

# On ajoute un 'User-Agent' pour ne pas être bloqué par le site
headers = {'User-Agent': 'Mozilla/5.0'}
response = requests.get(url, headers=headers)

# Lecture des tableaux présents sur la page
tables = pd.read_html(response.text)

# Généralement, le tableau principal est le premier ou le deuxième
# On l'enregistre dans votre dossier habituel
df_pib = tables[0]
df_pib.to_csv(r'C:\Users\freew\Desktop\Base de donnée\CI_PIB.csv', index=False)

print("Tableau extrait avec succès !")
```
